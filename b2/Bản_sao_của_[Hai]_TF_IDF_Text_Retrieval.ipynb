```python
# Download dataset: !gdown 1jhb2p2DlAwsDo_vEWICTrNh3mUuXd-cw6
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer

vi_data_df = pd.read_csv("./vi_text_retrieval.csv")
context = vi_data_df['text']
context = [doc.lower() for doc in context]
```

## **2. Tìm kiếm dựa trên TF-IDF**

+ **Khởi tạo Vectorizer**: Sử dụng `TfidfVectorizer` từ `sklearn.feature_extraction.text` để chuyển đổi văn bản thành ma trận số. `TF-IDF` đo lường tầm quan trọng của từ trong tài liệu so với toàn bộ tập dữ liệu.
+ **Huấn luyện Vectorizer**: `Vectorizer` được fit và `transform` trên danh sách `context` (danh sách các văn bản đã chuẩn hóa). Kết quả là một ma trận `sparse`, sau đó được chuyển thành mảng **(toarray())** để dễ xử lý.

```python
def tfidf_search(question, tfidf_vectorizer, top_d=5):
    # Lowercasing before encoding
    question = question.lower()
    query_embeded = tfidf_vectorizer.transform([question]).toarray()[0]

    cosine_scores = cosine_similarity([query_embeded], tfidf_vectorizer.transform(context).toarray())[0]

    # Get top k cosine score and index its
    results = []

    for idx in cosine_scores.argsort()[-top_d:][::-1]:
        doc_score = {
            'id': idx,
            'cosine_score': cosine_scores[idx]
        }
        results.append(doc_score)

    return results
```

```python
tfidf_vectorizer = TfidfVectorizer()
context_embeded = tfidf_vectorizer.fit_transform(context).toarray()
```

```python
question = vi_data_df.iloc[0]['question']
results = tfidf_search(question, tfidf_vectorizer, top_d=5)
results
```
